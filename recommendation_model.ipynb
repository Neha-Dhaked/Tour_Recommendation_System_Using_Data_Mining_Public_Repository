```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
import pickle

# Load tour data
tour_data = pd.read_csv("rawTourData.csv")

# Use TF-IDF to vectorize the tour descriptions
tfidf = TfidfVectorizer(stop_words='english')
tour_data['description'] = tour_data['description'].fillna('')
tfidf_matrix = tfidf.fit_transform(tour_data['description'])

# Compute cosine similarity matrix
cosine_sim = cosine_similarity(tfidf_matrix, tfidf_matrix)

# Save model and data to be used in Flask
pickle.dump(tfidf, open("tfidf_vectorizer.pkl", "wb"))
pickle.dump(cosine_sim, open("cosine_similarity.pkl", "wb"))
tour_data.to_csv("processed_tour_data.csv", index=False)
```
